```python
import glob
import os
import re

from ase.data import chemical_symbols

print(os.getcwd())

with open(f"..{os.sep}source_directory.txt") as fp:
    src_directory = f"{fp.readline().strip().replace('/', os.sep)}"
print(src_directory)
with open(f"..{os.sep}target_directory.txt") as fp:
    trg_directory = f"{fp.readline().strip().replace('/', os.sep)}"
print(trg_directory)
```

```python
mime_type = "image"
pattern = f"{trg_directory}{os.sep}*.{mime_type}.decompressed.csv"
decompression_logfiles: list[str] = glob.glob(pattern)
for file in sorted(decompression_logfiles):
    if not file.rsplit(os.sep, 1)[1].startswith("063"):
        continue
    with open(file, encoding="utf-8") as fp:
        for line in fp.readlines():
            tokenize: list[str] = line.strip().split(";")
            if len(tokenize) == 5:
                if tokenize[2].startswith(src_directory):
                    # {re.findall(r'[A-Z][a-z]?', tokenize[2][len(src_directory):])}"
                    print(
                        f"{tokenize[2][len(src_directory) :]}, {set([p for p in chemical_symbols[1:] if p in tokenize[2][len(src_directory) :]])}"
                    )
```
