```python
import os

import bibtexparser

print(os.getcwd())
with open("apm_source_directory.txt") as fp:
    apm_src_directory = f"{fp.readline().strip().replace('/', os.sep)}"
print(f"apm: {apm_src_directory}")
with open("em_source_directory.txt") as fp:
    em_src_directory = f"{fp.readline().strip().replace('/', os.sep)}"
print(f"em: {em_src_directory}")

cross_ref: dict[str, set[str]] = {"apm": set(), "em": set()}

with open(f"{apm_src_directory}{os.sep}aaa_legacy_data.bib") as fp:
    apm_bib = bibtexparser.load(fp).entries_dict
    print(len(apm_bib))
    for key, ref in apm_bib.items():
        if key.startswith("D_"):
            if "doi" in ref:
                # print(f"{key}, {ref['doi']}")
                cross_ref["apm"].add(ref["doi"].lower())
    print(f"{len(cross_ref['apm'])}")
with open(f"{em_src_directory}{os.sep}aaa_legacy_data.bib") as fp:
    em_bib = bibtexparser.load(fp).entries_dict
    print(len(em_bib))
    for key, ref in em_bib.items():
        if key.startswith("D"):
            if "doi" in ref:
                # print(f"{key}, {ref['doi']}")
                cross_ref["em"].add(ref["doi"].lower())
    print(f"{len(cross_ref['em'])}")

in_apm_but_not_in_em = cross_ref["apm"] - cross_ref["em"]
# print(len(in_apm_but_not_in_em))
for entry in in_apm_but_not_in_em:
    print(entry)
in_em_but_not_in_apm = cross_ref["em"] - cross_ref["apm"]
print(len(in_em_but_not_in_apm))
```
