```python
# find empty and missing directories
for idx in range(1, 815):
    path = f"{src_directory}{os.sep}{get_project_id(f'{idx}')}"
    if not os.path.isdir(path):
        print(f"{path} directory missing!")
        continue
    if not os.listdir(path):
        print(f"{path} exists but has no content!")
```

```python
# find zenodo zip archives which have in the meantime been unpacked
for file_path in file_paths:
    if file_path.endswith(".zip"):
        if not os.path.isfile(f"{src_directory}{os.sep}{file_path}"):
            print(f"File {file_path} has an issue")
    else:
        if not os.path.isdir(f"{src_directory}{os.sep}{file_path}"):
            print(f"Dir {file_path} has an issue")
```

```python
# move zenodo
import os
import shutil

for jdx, file_path in enumerate(file_paths):
    if jdx < 2:
        continue
    zenodo = file_path[file_path.rfind("/") + 1 :].replace(".zip", "")
    print(f"{idx + jdx}, {file_path}, {zenodo}")
    target_path = f"{src_directory}{os.sep}{get_project_id(f'{idx + jdx}')}"
    source_path = f"{src_directory}{os.sep}{file_path}"
    print(f"{source_path}\n{target_path}")
    os.makedirs(target_path, exist_ok=True)
    shutil.move(source_path, target_path)
    print("moved")
    # break
```
